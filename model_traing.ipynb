```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split, GridSearchCV
import joblib
```

```python
dataframe["progress_score"] = (
    (dataframe["mental_scale"] + dataframe["Sleeping_hr"] + (dataframe["social_duration"]/10) + np.random.uniform(-3, 3, len(dataframe))) / 20
).clip(0, 1)

# Scale to %
dataframe["progress_percent"] = (dataframe["progress_score"] * 100).round(2)
dataframe.drop(columns=["progress"], inplace=True)

dataframe.columns.tolist()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# -------------------------
# 4) Scale numeric features
# -------------------------
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# 5) GridSearchCV (hyperparameter tuning)
# -------------------------
from sklearn.multioutput import MultiOutputRegressor
param_grid = {
    'estimator__n_estimators': [100, 200],
    'estimator__max_depth': [5, 10, None],
    'estimator__min_samples_split': [2, 5],
    'estimator__min_samples_leaf': [1, 2]
}

grid = GridSearchCV(
    MultiOutputRegressor(RandomForestRegressor(random_state=42)),
    param_grid,
    scoring="r2",
    cv=3,
    verbose=2,
    n_jobs=-1
)
```

```python
print("Starting GridSearchCV (this can take some time)...")
grid.fit(X_train_scaled, y_train)

print("✅ Best Parameters:", grid.best_params_)
print("✅ Best R² Score (CV):", grid.best_score_)

# -------------------------
```

```python
best_model = grid.best_estimator_
# retrain on the full training set (GridSearchCV already refit by default, but this ensures we are explicit)
best_model.fit(X_train_scaled, y_train)

# -------------------------
# 7) Evaluate on test set
# -------------------------
y_pred = best_model.predict(X_test_scaled)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print("\n📊 FINAL TEST PERFORMANCE:")
print(f"MAE: {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²: {r2:.4f}")
```

```python
# Show a small table of actual vs predicted
# Create a DataFrame for predicted results
test_results = pd.DataFrame({
    "Actual Progress Score": y_test.iloc[:, 0].reset_index(drop=True),
    "Actual Progress Percent": y_test.iloc[:, 1].reset_index(drop=True),
    "Predicted Progress Score": y_pred[:, 0],
    "Predicted Progress Percent": y_pred[:, 1]
})

print("\n🔍 Sample Predictions:")
print(test_results.head(10))

```

```python
# 8) Line chart: Actual vs Predicted (sorted)
# -------------------------
# Sort by actual value for a clearer line visualization

import matplotlib.pyplot as plt

# Sort values by Actual Progress Score
sorted_idx = test_results["Actual Progress Score"].argsort().values
actual_sorted = test_results["Actual Progress Score"].values[sorted_idx]
pred_sorted = test_results["Predicted Progress Score"].values[sorted_idx]

plt.figure(figsize=(12, 5))
plt.plot(actual_sorted, label="Actual Progress Score", marker="o")
plt.plot(pred_sorted, label="Predicted Progress Score", marker="x")
plt.title("Actual vs Predicted Progress Score (Line Chart)")
plt.xlabel("Test Samples (sorted by actual score)")
plt.ylabel("Progress Score")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("progress_score_line_chart.png")
print("📊 Line chart saved to progress_score_line_chart.png")
plt.show()
```

```python
import numpy as np
import pandas as pd

def predict_progress( age, gender, disorder, mentalscale, sleeping_hr, duration_interaction):
    """
    Predicts both progress_score (0–1) and progress_percent (0–100)
    for a given user's input data.
    """

    # --- Encode gender ---
    gender_Male = 1 if gender.lower() == "male" else 0

    # --- Initialize all disorder flags ---
    disorders = {
        "disorder_Arthritis": 0,
        "disorder_Autism Spectrum Disorder": 0,
        "disorder_Blindness": 0,
        "disorder_Dementia": 0,
        "disorder_Depression": 0,
        "disorder_Diabetes": 0,
        "disorder_Down Syndrome": 0,
        "disorder_Epilepsy": 0,
        "disorder_OCD": 0,
    }

    # --- Activate selected disorder(s) ---
    if isinstance(disorder, list):
        for d in disorder:
            key = f"disorder_{d}"
            if key in disorders:
                disorders[key] = 1
    else:
        key = f"disorder_{disorder}"
        if key in disorders:
            disorders[key] = 1

    # --- Build feature dict (exclude user_ID because model didn’t train on it) ---
    features = {
        "age": age,
        "social_duration": duration_interaction,
        "mental_scale": mentalscale,
        "Sleeping_hr": sleeping_hr,
        "gender_Male": gender_Male,
        **disorders
    }

    # --- Convert to DataFrame ---
    data_f = pd.DataFrame([features])

    # --- Scale input ---
    data_scaled = scaler.transform(data_f)

    # --- Predict both outputs (model outputs [progress_score, progress_percent]) ---
    predicted = best_model.predict(data_scaled)[0]
    progress_score = round(float(predicted[0]), 3)
    progress_percent = round(float(predicted[1]), 2)

    # --- Print results neatly ---
    
    print(f"📊 Predicted Progress Score: {progress_score}")
    print(f"📈 Predicted Progress Percent: {progress_percent}%")

    # --- Return values as dict for flexibility ---
    return {
        
        "progress_score": progress_score,
        "progress_percent": progress_percent
    }
sample_result = predict_progress(
    #user_ID=501,
    age=11,
    gender="malw",
    disorder="Blindness",
    mentalscale=5,
    sleeping_hr=9.5,
    duration_interaction=6
)


```
